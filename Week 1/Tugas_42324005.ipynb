```python
import numpy as np

# Membuat matriks random 5x5
# Nilai random dari 0 sampai 100
matriks = np.random.randint(0, 101, size=(5, 5))

print("Matriks 5x5:")
print(matriks)

# Nilai maksimum
nilai_maksimum = np.max(matriks)

# Nilai minimum
nilai_minimum = np.min(matriks)

# Nilai rata-rata
nilai_rata_rata = np.mean(matriks)

print("\nNilai maksimum :", nilai_maksimum)
print("Nilai minimum  :", nilai_minimum)
print("Nilai rata-rata:", nilai_rata_rata)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat range x dan y
x = np.linspace(-2, 2, 1000)
y = np.linspace(-2, 2, 1000)

# Membuat meshgrid
X, Y = np.meshgrid(x, y)

# Persamaan:
# x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4
# - x^2y^3 - 6x^2y^2 + 3x^2
# + y^6 - 3y^4 + 3y^2 = 1

F = (
    X**6
    + 3*X**4*Y**2
    - 3*X**4
    + 3*X**2*Y**4
    - X**2*Y**3
    - 6*X**2*Y**2
    + 3*X**2
    + Y**6
    - 3*Y**4
    + 3*Y**2
)

# Membuat grafik F = 1
plt.figure(figsize=(8, 8))

plt.contour(X, Y, F, levels=[1])

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Grafik 2D Persamaan")

plt.grid(True)
plt.axis("equal")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat range x dan y
x = np.linspace(-4, 4, 200)
y = np.linspace(-4, 4, 200)

# Membuat meshgrid
X, Y = np.meshgrid(x, y)

# Persamaan:
# z = (5y/4 - |x|)^2 + x^2

Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

# Membuat grafik 3D
fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection="3d")

# Plot permukaan
permukaan = ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis",
    edgecolor="none"
)

# Label sumbu
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

# Judul
ax.set_title(r"$z = (5y/4 - |x|)^2 + x^2$")

# Colorbar
fig.colorbar(
    permukaan,
    ax=ax,
    shrink=0.6,
    label="Nilai Z"
)

plt.show()
```
