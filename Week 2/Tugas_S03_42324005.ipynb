```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path = "diabetes.csv"
df2 = pd.read_csv(path)

df2
```

```python
df2.info()
```

```python
columns = ['Glucose', 'BloodPressure', 'BMI']

for col in columns:
    sns.boxplot(x=df2[col])
    plt.title("Boxplot " + col)
    plt.show()
```

```python
q1 = np.percentile(df2['Glucose'], 25)
q3 = np.percentile(df2['Glucose'], 75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
```

```python
q1 = np.percentile(df2['BloodPressure'], 25)
q3 = np.percentile(df2['BloodPressure'], 75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
```

```python
q1 = np.percentile(df2['BMI'], 25)
q3 = np.percentile(df2['BMI'], 75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
```

```python
def remove_outliers(df, columns):
    df_cleaned = df.copy()

    for col in columns:
        q1 = np.percentile(df_cleaned[col], 25)
        q3 = np.percentile(df_cleaned[col], 75)

        iqr = q3 - q1

        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)

        df_cleaned = df_cleaned[
            (df_cleaned[col] >= lower_bound) &
            (df_cleaned[col] <= upper_bound)
        ]

    return df_cleaned
```

```python
clean_data = remove_outliers(df2, ['Glucose', 'BloodPressure', 'BMI'])

clean_data
```

```python
print("Jumlah data sebelum handling:", len(df2))
print("Jumlah data setelah handling:", len(clean_data))
```

```python
for col in columns:
    sns.boxplot(x=clean_data[col])
    plt.title("Boxplot " + col + " Setelah Handling Outliers")
    plt.show()
```
